```python
import numpy as np
import jax.numpy as jnp
from jax import grad, jit, vmap, device_put
import time

# Create a large 5000x5000 matrix
size = 5000
x_cpu = np.random.normal(size=(size, size)).astype(np.float32)

# --- Benchmark NumPy ---
start = time.time()
result_np = np.dot(x_cpu, x_cpu)
print(f"NumPy (CPU) Time: {time.time() - start:.4f}s")

# --- Benchmark JAX (Accelerated) ---
# Move data to the Accelerator (GPU/TPU)
x_jax = device_put(jnp.array(x_cpu))

# Define a JIT-compiled function
@jit
def fast_matmul(a):
    return jnp.matmul(a, a)

# Warm-up (JAX compiles on the first run)
_ = fast_matmul(x_jax).block_until_ready()

start = time.time()
result_jax = fast_matmul(x_jax).block_until_ready()
print(f"JAX (Accelerator) Time: {time.time() - start:.4f}s")
```
